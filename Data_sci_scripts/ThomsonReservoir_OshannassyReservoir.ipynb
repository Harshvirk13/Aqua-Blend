```python
import pandas as pd
```

```python
# Read the original CSV file
df = pd.read_csv("Thomson and O'Shannassy Reservoir.csv")

# Show the first five rows
df.head()
```

```python
# Display all column names
print(df.columns)
```

```python
# Convert the Datetime column into proper date format
df["Datetime"] = pd.to_datetime(
    df["Datetime"],
    format="%Y/%m/%d %H:%M:%S",
    errors="coerce"
)

# Convert the Value column into numeric format
df["Value"] = pd.to_numeric(
    df["Value"],
    errors="coerce"
)
```

```python
# Replace unclear site names with proper reservoir names
df["Name"] = df["Name"].replace({
    "THOMSON @ THOMSON AD": "Thomson Reservoir",
    "YAOSH0127": "O'Shannassy Reservoir"
})

# Check the updated names
print(df["Name"].unique())
```

```python
# Remove records marked as NV
df = df[df["Qualifier"] != "NV"]

# Remove rows with missing important information
df = df.dropna(
    subset=["Name", "Datetime", "Parameter", "Value"]
)

print("Remaining rows:", len(df))
```

```python
# Keep only the columns needed for the project
clean_df = df[
    [
        "Site ID",
        "Name",
        "Datetime",
        "Parameter",
        "Value",
        "Unit of Measurement",
        "Quality"
    ]
].copy()
```

```python
# Rename columns to make them easier to use in Python and Supabase
clean_df = clean_df.rename(columns={
    "Site ID": "site_id",
    "Name": "source_name",
    "Datetime": "measurement_datetime",
    "Parameter": "parameter",
    "Value": "value",
    "Unit of Measurement": "unit",
    "Quality": "quality_code"
})

clean_df.head()
```

```python
# Sort the data by reservoir, date and parameter
clean_df = clean_df.sort_values(
    by=[
        "source_name",
        "measurement_datetime",
        "parameter"
    ]
)
```

```python
# Save the cleaned data into a new CSV file
clean_df.to_csv(
    "thomson_oshannassy_clean.csv",
    index=False
)

print("Data cleaning completed successfully.")
print("Number of cleaned rows:", len(clean_df))
```
